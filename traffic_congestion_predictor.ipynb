# Lagos Traffic Congestion Predictor

**3MTT AI/ML Project | AI-16**

This notebook develops a machine-learning classification model that predicts traffic congestion level as **Low, Medium, or High** using historical smart-mobility traffic observations.

> **Important scope note:** The 3MTT brief identifies Lagos commuters as the use case, but the selected public dataset is not official Lagos traffic data. The project therefore demonstrates the ML methodology as a proof of concept and should not be presented as a model validated on real Lagos traffic.

## 1. Problem Statement

Traffic congestion creates uncertainty for commuters. The objective is to build a supervised machine-learning model that estimates congestion level from historical traffic and contextual information.

### Objectives
- Inspect and clean the historical traffic dataset.
- Explore patterns associated with congestion.
- Engineer time-based features.
- Compare multiple classification algorithms.
- Evaluate the best model using accuracy, precision, recall, F1-score, and a confusion matrix.
- Save the trained model for reuse.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_recall_fscore_support
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
import joblib

df = pd.read_csv("data/smart_mobility_dataset.csv")
df.head()
```

```python
df.shape, df.columns.tolist()
```

```python
df.info()
print("\nMissing values:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Data Preparation

The dataset contains 5,000 records and 15 original columns. The target variable is `Traffic_Condition`.

The original dataset contains no missing values and no duplicate rows.

```python
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

df["Hour"] = df["Timestamp"].dt.hour
df["DayOfWeek"] = df["Timestamp"].dt.dayofweek
df["DayName"] = df["Timestamp"].dt.day_name()
df["IsWeekend"] = (df["DayOfWeek"] >= 5).astype(int)

# Peak periods are defined as common morning/evening commuting windows.
df["PeakPeriod"] = df["Hour"].isin([7, 8, 9, 16, 17, 18, 19]).astype(int)

df[["Timestamp","Hour","DayOfWeek","IsWeekend","PeakPeriod"]].head()
```

## 3. Exploratory Data Analysis

The target is imbalanced: High congestion is the majority class while Low congestion is the minority class. Therefore, accuracy alone is not sufficient. Macro-averaged precision, recall and F1-score are also reported.

## 4. Target Leakage / Label Construction Investigation

Before modelling, we investigate whether the target appears to be strongly determined by direct traffic measurements.

A shallow decision tree can reproduce the target at high accuracy using variables such as vehicle count, speed, occupancy and accident status. This indicates that the supplied congestion label is strongly rule-like or derived from traffic measurements.

This is an important limitation. We therefore report model performance transparently rather than treating a very high score as proof of real-world Lagos performance.

```python
from sklearn.tree import DecisionTreeClassifier

leak_features = ["Vehicle_Count", "Traffic_Speed_kmh", "Road_Occupancy_%", "Accident_Report"]
X_leak = df[leak_features]
y = df["Traffic_Condition"]

Xtr, Xte, ytr, yte = train_test_split(
    X_leak, y, test_size=.20, stratify=y, random_state=42
)

for depth in [1,2,3,4,5,8]:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=42)
    tree.fit(Xtr, ytr)
    pred = tree.predict(Xte)
    print(f"depth={depth}: accuracy={accuracy_score(yte,pred):.4f}")
```

## 5. Feature Engineering and Model Training

The final modelling table includes direct traffic measurements plus contextual and time-derived variables.

Categorical variables are one-hot encoded and numerical variables are standardized inside a scikit-learn pipeline. This prevents preprocessing leakage between training and test data.

```python
features = [
    "Vehicle_Count", "Traffic_Speed_kmh", "Road_Occupancy_%",
    "Traffic_Light_State", "Weather_Condition", "Accident_Report",
    "Sentiment_Score", "Ride_Sharing_Demand", "Parking_Availability",
    "Emission_Levels_g_km", "Energy_Consumption_L_h",
    "Hour", "DayOfWeek", "IsWeekend", "PeakPeriod"
]

X = df[features]
y = df["Traffic_Condition"]

categorical = X.select_dtypes(include="object").columns.tolist()
numeric = [c for c in X.columns if c not in categorical]

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=.20, stratify=y, random_state=42
)
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=2500, class_weight="balanced"),
    "Decision Tree": DecisionTreeClassifier(max_depth=5, class_weight="balanced", random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=300, max_depth=12, class_weight="balanced",
        random_state=42, n_jobs=-1
    )
}

results = []
fitted_models = {}

for name, model in models.items():
    pipe = Pipeline([("preprocessor", preprocessor), ("model", model)])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)

    precision, recall, f1, _ = precision_recall_fscore_support(
        y_test, pred, average="macro", zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": f1
    })
    fitted_models[name] = (pipe, pred)

results_df = pd.DataFrame(results).sort_values("Macro F1", ascending=False)
results_df
```

## 6. Evaluation

The model is evaluated using:
- Accuracy
- Macro precision
- Macro recall
- Macro F1-score
- Confusion matrix

Macro averages are important because the classes are imbalanced.

```python
final_model = fitted_models["Random Forest"][0]
final_pred = fitted_models["Random Forest"][1]

print(classification_report(
    y_test, final_pred,
    labels=["Low","Medium","High"],
    digits=4,
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, final_pred, labels=["Low","Medium","High"])

plt.figure(figsize=(6,5))
plt.imshow(cm)
plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(3), ["Low","Medium","High"])
plt.yticks(range(3), ["Low","Medium","High"])

for i in range(3):
    for j in range(3):
        plt.text(j, i, cm[i,j], ha="center", va="center")

plt.tight_layout()
plt.show()
```

## 7. Save the Trained Model

The preprocessing and model are stored together in one scikit-learn pipeline so that future predictions use the same transformations applied during training.

```python
joblib.dump(final_model, "models/traffic_congestion_model.joblib")
print("Saved: models/traffic_congestion_model.joblib")
```

## 8. Example Prediction

The following example shows how the saved model can be used for a hypothetical traffic observation. This is a demonstration prediction, not a live Lagos traffic forecast.

```python
sample = pd.DataFrame([{
    "Vehicle_Count": 180,
    "Traffic_Speed_kmh": 42,
    "Road_Occupancy_%": 65,
    "Traffic_Light_State": "Red",
    "Weather_Condition": "Clear",
    "Accident_Report": 0,
    "Sentiment_Score": 0.1,
    "Ride_Sharing_Demand": 60,
    "Parking_Availability": 40,
    "Emission_Levels_g_km": 120,
    "Energy_Consumption_L_h": 8,
    "Hour": 17,
    "DayOfWeek": 2,
    "IsWeekend": 0,
    "PeakPeriod": 1
}])

prediction = final_model.predict(sample)[0]
probabilities = final_model.predict_proba(sample)[0]

print("Predicted congestion:", prediction)
print("Model confidence:", f"{probabilities.max():.1%}")
```
